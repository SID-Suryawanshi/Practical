```python
import pandas as pd
import numpy as np

from sklearn.svm import SVC
from sklearn.neighbors import KNeighborsClassifier
from sklearn import metrics
from sklearn.preprocessing import scale
from sklearn.model_selection import train_test_split

# Import dataset
df = pd.read_csv("emails.csv")

# Check dataset
df

# First five rows
df.head()

# Get size of dataset
df.shape

# Get data types
df.dtypes

# Get column names
df.columns

# Generate summary statistics
df.describe()

# Check null values
s = df.isnull().sum()

s[s > 0]

df.isnull().any()

# Drop unnecessary rows/columns containing null values
df.dropna(inplace=True)

# Drop Email No. column
df.drop(['Email No.'], axis=1, inplace=True)

# Check first five rows
df.head()

# Check columns
df.columns

# Check unique prediction values
df.Prediction.unique()

# Replace prediction values
df['Prediction'] = df['Prediction'].replace({
    0: 'Not spam',
    1: 'Spam'
})

df.columns

# Display dataset
df

# Separate features and target
X = df.drop(columns='Prediction', axis=1)
Y = df['Prediction']

# Scale features
X = scale(X)

# Split into training and testing data
X_train, X_test, y_train, y_test = train_test_split(
    X,
    Y,
    test_size=0.3,
    random_state=42
)

# KNN Classifier
knn = KNeighborsClassifier(n_neighbors=7)

knn.fit(X_train, y_train)

y_pred = knn.predict(X_test)

print("Prediction: \n")
print(y_pred)

# KNN Accuracy
M = metrics.accuracy_score(y_test, y_pred)
print("KNN accuracy: ", M)

# KNN Confusion Matrix
C = metrics.confusion_matrix(y_test, y_pred)
print("Confusion matrix: ", C)

# SVM Classifier
model = SVC(C=1)

# Fit model
model.fit(X_train, y_train)

# Predict
y_pred = model.predict(X_test)

# SVM Confusion Matrix
metrics.confusion_matrix(
    y_true=y_test,
    y_pred=y_pred
)

# SVM Accuracy
print(
    "SVM accuracy = ",
    metrics.accuracy_score(y_test, y_pred)
)
```
